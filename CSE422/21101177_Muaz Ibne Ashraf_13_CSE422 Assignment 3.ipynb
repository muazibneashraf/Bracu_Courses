```python
import random as rand
import math

class TreeNode:
    def __init__(self, value=None):
        self.value = value
        self.children = []

def build_game_tree(depth, branching_factor):
    root = TreeNode()
    if depth == 0:
        return root

    for _ in range(branching_factor):
        child = build_game_tree(depth - 1, branching_factor)
        root.children.append(child)

    return root

def strength(x):
    return math.log2(x + 1) + x / 10

def assign_leaf_values(node, depth, max_depth, maxV, minV):
    if depth == max_depth:
        i = rand.randint(0, 1)
        random_factor = rand.randint(1, 10) / 10
        utility = strength(maxV) - strength(minV) + ((-1) ** i) * random_factor
        node.value = round(utility, 2)
    else:
        for child in node.children:
            assign_leaf_values(child, depth + 1, max_depth, maxV, minV)

def alpha_beta_pruning(node, depth, alpha, beta, is_max):
    if depth == 0 or node.value is not None:
        return node.value

    if is_max:
        value = float('-inf')
        for child in node.children:
            value = max(value, alpha_beta_pruning(child, depth - 1, alpha, beta, False))
            alpha = max(alpha, value)
            if beta <= alpha:
                break
        return value
    else:
        value = float('inf')
        for child in node.children:
            value = min(value, alpha_beta_pruning(child, depth - 1, alpha, beta, True))
            beta = min(beta, value)
            if beta <= alpha:
                break
        return value

def play_chess_games():
    start_player = int(input("Enter starting player for game 1 (0 for Carlsen, 1 for Caruana): "))
    carlsen_strength = float(input("Enter base strength for Carlsen: "))
    caruana_strength = float(input("Enter base strength for Caruana: "))

    players = ["Magnus Carlsen", "Fabiano Caruana"]
    wins = [0, 0]
    draws = 0

    for game in range(1, 5):
        if game % 2 == 1:
            max_player = start_player
            min_player = 1 - start_player
            maxV = carlsen_strength if max_player == 0 else caruana_strength
            minV = caruana_strength if min_player == 0 else carlsen_strength
        else:
            max_player = 1 - start_player
            min_player = start_player
            maxV = caruana_strength if max_player == 1 else carlsen_strength
            minV = carlsen_strength if min_player == 1 else caruana_strength


        game_tree = build_game_tree(5, 2)
        assign_leaf_values(game_tree, 0, 5, maxV, minV)


        result = alpha_beta_pruning(game_tree, 5, float('-inf'), float('inf'), True)


        if result > 0:
            winner = max_player
            wins[max_player] += 1
            win_type = "Max"
        elif result < 0:
            winner = min_player
            wins[min_player] += 1
            win_type = "Min"
        else:
            winner = None
            draws += 1
            win_type = "Draw"


        if winner is not None:
            print(f"Game {game} Winner: {players[winner]} ({win_type}) (Utility value: {result:.2f})")
        else:
            print(f"Game {game} Result: Draw (Utility value: {result:.2f})")


    print("Overall Results:")
    print(f"Magnus Carlsen Wins: {wins[0]}")
    print(f"Fabiano Caruana Wins: {wins[1]}")
    print(f"Draws: {draws}")

    if wins[0] > wins[1]:
        print("Overall Winner: Magnus Carlsen")
    elif wins[1] > wins[0]:
        print("Overall Winner: Fabiano Caruana")
    else:
        print("Overall Winner: Draw")


if __name__ == "__main__":
    play_chess_games()
```
