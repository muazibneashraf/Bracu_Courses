```python
import random as rnd
f=open("/content/inp_1","r")
num=int(f.readline())
inp=[]
for i in range(num):
  s,d=f.readline().split()
  if s=="l":
    inp+=[0-int(d)]
  elif s=="d":
    inp+=[int(d)]
print(inp)


def register(num):
  while True:
    reg=[rnd.choice(["0","1"]) for i in range(num)]
    if reg!=["0"]*num:
      break
  return reg

def fitness(reg,inp):
  res=0
  for i in range(len(reg)):
    if reg[i]=="1":
      res+=inp[i]
  return res



def crossover(reg1,reg2):
  crs_point=rnd.randint(1, len(reg1) - 1)
  child1 = reg1[:crs_point] + reg2[crs_point:]
  child2 = reg1[:crs_point] + reg2[crs_point:]
  return child1, child2

def mutation(reg):
  mut_point=rnd.randint(1, len(reg) - 1)
  while True:
    if reg[mut_point]=="1":
      reg[mut_point]="0"
    else:
      reg[mut_point]="1"
    if reg!=["0"]*num:
      break
  return reg

def genetic(num,inp):
  reg1=register(num)
  reg2=register(num)
  popu=[reg1,reg2]
  for i in range(3000):
    fit=[fitness(reg1,inp),fitness(reg2,inp)]
    for i in range(len(fit)):
      if fit[i]==0:
        return popu[i]

    child1,child2=crossover(reg1,reg2)
    reg1=mutation(child1)
    reg2=mutation(child2)
    popu=[reg1,reg2]
  return [-1]

res=genetic(num,inp)
x=""
for i in res:
  x+=f"{i}"
print(x)
```
