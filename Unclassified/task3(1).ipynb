```python
inp=open('input1_A.txt','r')
out=open('output1_A.txt','w')
V,E=tuple(map(int,inp.readline().split()))
matrix = [[] for row in range(0, V+1)]

for i in range(E):
   a,b=tuple(map(int,inp.readline().split()))
   matrix[a].append(b)
print(matrix)


cycle=''
def topologicalSort_dfs(graph,v):
    vis=['white']*(v+1)
    stack=[]
    for i in range(1,v+1):
       if vis[i]=='white':
          top_dfs_visit(graph,i,vis,stack)
    return stack

def top_dfs_visit(graph,u,vis,stack):
    vis[u]='gray'
    for v in graph[u]:

       if vis[v]=='gray':
          global cycle
          cycle='yes'
          print('h')
       elif vis[v]=='white':
          top_dfs_visit(graph,v,vis,stack)

    vis[u]='black'
    stack.append(u)

x=topologicalSort_dfs(matrix,V)
if cycle!='yes':
   for i in range(len(x)-1,-1,-1):
      out.write(str(x[i])+' ')
else:
   print('Impossible',file=out)
```
