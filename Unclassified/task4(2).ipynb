```python
input = open('/content/input4.txt', 'r')
output = open('/content/output4.txt', 'w')

data_num = int(input.readline())
data_list = input.readlines()
start = []
stop = []
time = []

for i in data_list:
  a = i.split()
  start.append(a[0])
  stop.append(a[4])
  time.append(a[-1])
for i in range(data_num):
  for j in range(0, data_num - i - 1):
    if start[j] > start[j + 1] or (start[j] == start[j + 1] and time[j] < time[j + 1]):
      start[j], start[j + 1] = start[j + 1], start[j]
      stop[j], stop[j + 1] = stop[j + 1], stop[j]
      time[j], time[j + 1] = time[j + 1], time[j]

for i in range(data_num):
  result = start[i] + ' will departure for ' + stop[i] + ' at ' + time[i] + '\n'
  output.write(str(result))
output.close()
```
