## 1. Introduction
This project focuses on predicting whether a customer will churn using machine learning models. Customer churn prediction is essential for businesses to improve customer retention strategies. The primary goal is to analyze consumer data, preprocess it, and apply classification models such as Logistic Regression, Decision Tree, Neural Network, Naive Bayes, and KNN.

## 2. Dataset Description
We begin by loading and exploring the dataset.

```python
import pandas as pd
df = pd.read_csv("/content/consumer_classification_dataset.csv")
df.info()
df.describe()
```

### Heatmap for Feature Correlation

```python
import seaborn as sns
import matplotlib.pyplot as plt

eda_df = df.dropna()
eda_encoded = eda_df.copy()
eda_encoded["Gender"] = eda_encoded["Gender"].astype("category").cat.codes
eda_encoded["Marital_Status"] = eda_encoded["Marital_Status"].astype("category").cat.codes
eda_encoded["Device_Used"] = eda_encoded["Device_Used"].astype("category").cat.codes

plt.figure(figsize=(10, 6))
sns.heatmap(eda_encoded.corr(), annot=True, cmap='coolwarm')
plt.title("Feature Correlation Heatmap")
plt.show()
```

### Imbalanced Dataset Check

```python
sns.countplot(data=df, x='Churn')
plt.title("Class Distribution of Target Feature 'Churn'")
plt.show()
```

### 3. Dataset Preprocessing
We handle missing values, encode categorical features, and scale numerical values.

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import LabelEncoder, StandardScaler

data = df.copy()
num_cols = ['Age', 'Income', 'Credit_Score']
imputer = SimpleImputer(strategy='mean')
data[num_cols] = imputer.fit_transform(data[num_cols])

cat_cols = ['Gender', 'Marital_Status', 'Device_Used']
encoders = {col: LabelEncoder() for col in cat_cols}
for col in cat_cols:
    data[col] = encoders[col].fit_transform(data[col])

scaler = StandardScaler()
X = data.drop('Churn', axis=1)
y = data['Churn']
X_scaled = scaler.fit_transform(X)
```

### 4. Dataset Splitting

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42, stratify=y)
X_train.shape, X_test.shape
```

### 5. Model Training & Testing
We'll apply Logistic Regression, Decision Tree, Neural Network, KNN, and Naive Bayes.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, roc_auc_score, roc_curve

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "Decision Tree": DecisionTreeClassifier(),
    "Neural Network": MLPClassifier(max_iter=1000),
    "KNN": KNeighborsClassifier(),
    "Naive Bayes": GaussianNB()
}

results = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    y_prob = model.predict_proba(X_test)[:, 1] if hasattr(model, "predict_proba") else None

    results[name] = {
        "Accuracy": accuracy_score(y_test, y_pred),
        "Confusion Matrix": confusion_matrix(y_test, y_pred),
        "Classification Report": classification_report(y_test, y_pred),
        "ROC AUC": roc_auc_score(y_test, y_prob) if y_prob is not None else "N/A"
    }

results
```

### 6. Model Selection & Comparison

```python
import matplotlib.pyplot as plt

model_names = list(results.keys())
accuracies = [results[m]["Accuracy"] for m in model_names]

plt.figure(figsize=(8, 5))
sns.barplot(x=model_names, y=accuracies)
plt.title("Model Accuracy Comparison")
plt.ylabel("Accuracy")
plt.ylim(0, 1)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 6))
for name, model in models.items():
    if hasattr(model, "predict_proba"):
        y_prob = model.predict_proba(X_test)[:, 1]
        fpr, tpr, _ = roc_curve(y_test, y_prob)
        plt.plot(fpr, tpr, label=f"{name} (AUC = {results[name]['ROC AUC']:.2f})")
plt.plot([0, 1], [0, 1], 'k--')
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve Comparison")
plt.legend()
plt.show()
```

### 7. Conclusion
All models performed well, with Decision Tree and Neural Network yielding the highest accuracy. Feature preprocessing significantly improved performance. The dataset’s slight imbalance was handled effectively using stratified splitting.
