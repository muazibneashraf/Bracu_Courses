```python
file1 = open("/content/input1b.txt","r")
file2 = open("/content/output1b.txt","w")
t = file1.readline()
for i in range(int(t)):
  a = file1.readline().strip()
  operation = a[10:]
  #print(operation)
  list1 = operation.split(" ")
  #print(list1)
  res = 0

  if list1[1] == "+":
    res = int(list1[0]) + int(list1[2])
  elif list1[1] == "-":
    res = int(list1[0]) - int(list1[2])
  if list1[1] == "/":
    res = int(list1[0]) / int(list1[2])
  elif list1[1] == "*":
    res = int(list1[0]) * int(list1[2])
  file2.writelines(f"The result of {operation} is {res}\n")
file1.close()
file2.close()
```
