```python
import heapq

def dijkstra(graph, start):
    distances = {node: float('inf') for node in graph}
    distances[start] = 0
    heap = [(0, start)]

    while heap:
        current_distance, current_node = heapq.heappop(heap)

        if current_distance > distances[current_node]:
            continue

        for neighbor, danger in graph[current_node].items():
            distance = max(current_distance, danger)
            if distance < distances[neighbor]:
                distances[neighbor] = distance
                heapq.heappush(heap, (distance, neighbor))

    return distances

def find_safest_path(graph, N):
    distances = dijkstra(graph, 1)
    return distances[N]

with open("input3.txt", "r") as inp_file, open("output3.txt", "w") as out_file:
    N, M = map(int, inp_file.readline().split())
    graph = {i: {} for i in range(1, N+1)}

    for _ in range(M):
        u, v, w = map(int, inp_file.readline().split())
        graph[u][v] = w

    safest_danger_level = find_safest_path(graph, N)
    if safest_danger_level == float('inf'):
        out_file.write("Impossible\n")
    else:
        out_file.write(str(safest_danger_level) + "\n")
```
