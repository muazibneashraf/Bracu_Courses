```python
# -*- coding: utf-8 -*-
"""task2.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/12cyR4ACvKr1UHW7OLwjuaZ9U5q72oRt6
"""

inp_file = open('/content/input2.txt', 'r')
out_file = open('/content/output2.txt', 'w')
x = inp_file.readlines()
N = int(x[0])
A = list(map(int, x[1].split()))


def findMax(arr, left, right):
    if left == right:
        return arr[left]

    mid = (left + right) // 2

    maxLeft = findMax(arr, left, mid)
    maxRight = findMax(arr, mid + 1, right)

    return max(maxLeft, maxRight)


maxValue = str(findMax(A, 0, N - 1))
out_file.write(maxValue)
out_file.close()
```

```python
# -*- coding: utf-8 -*-
"""task3.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/12cyR4ACvKr1UHW7OLwjuaZ9U5q72oRt6
"""

def alien_competition(arr):
  count = 0
  i = 0
  j = len(arr) - 1
  while i<= j:
    if arr[i] > arr[j]:
      count += 1
      j -= 1
    else:
      if i == j:
        i+= 1
        j = len(arr) - 1
      else:
        j-= 1
  return count

with open("input3.txt", "r") as inp_file, open("output3.txt", "w") as out_file:
  n = int(inp_file.readline())
  arr = list(map(int, inp_file.readline().split(" ")))
  count = alien_competition(arr)
  out_file.write(f"{count}")
```

```python
# -*- coding: utf-8 -*-
"""task4.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/12cyR4ACvKr1UHW7OLwjuaZ9U5q72oRt6
"""

def find_max_sum_squared(n, arr):
  max_sum = float('-inf')
  max_value = float('-inf')
  i = 0
  j = 1
  while j < n:
    current_sum = arr[i] + arr[j] ** 2
    if current_sum > max_sum:
        max_sum = current_sum
        max_value = arr[i]
    if arr[j] <= max_value:
        j += 1
    else:
        i += 1
        if arr[i] > max_value:
            max_value = arr[i]
            j = i + 1
  return max_sum

with open("input4.txt", "r") as inp_file, open("output4.txt", "w") as out_file:
  n = int(inp_file.readline())
  arr = list(map(int, inp_file.readline().split(" ")))
  max_sum = find_max_sum_squared(n, arr)
  out_file.write(f"{max_sum}")
```

```python
# -*- coding: utf-8 -*-
"""task5.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/12cyR4ACvKr1UHW7OLwjuaZ9U5q72oRt6
"""

inp_file = open('/content/input5.txt', 'r')
out_file = open('/content/output5.txt', 'w')

lines = inp_file.readlines()
n = int(lines[0])
nums = list(map(int, lines[1].split()))

def quick_sort(arr):
    if len(arr) <= 1:
        return arr

    pivot = arr[0]
    left = []
    middle = []
    right = []

    for x in arr:
        if x < pivot:
            left.append(x)
        elif x == pivot:
            middle.append(x)
        else:
            right.append(x)

    return quick_sort(left) + middle + quick_sort(right)


sorted_nums = quick_sort(nums)
output_str = ' '.join(map(str, sorted_nums))
out_file.write(output_str)
inp_file.close()
out_file.close()
```

```python
# -*- coding: utf-8 -*-
"""task6.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/12cyR4ACvKr1UHW7OLwjuaZ9U5q72oRt6
"""

def partition(lst,low,high):
    index = low
    pivot = lst[index]
    while low < high:
        while low < len(lst) and lst[low] <= pivot:
            low += 1
        while lst[high] > pivot:
            high -= 1
        if low < high:
            lst[low], lst[high] = lst[high], lst[low]
    lst[high], lst[index] = lst[index], lst[high]
    return high


def kth_smallest_value(arr, K):
    low = 0
    high = len(arr) - 1
    while low <= high:
        pivot_index = partition(arr, low, high)
        if pivot_index == K - 1:
            return arr[pivot_index]
        elif pivot_index > K - 1:
            high = pivot_index - 1
        else:
            low = pivot_index + 1

with open("input6.txt", "r") as inp_file, open("output6.txt", "w") as out_file:
    n = int(inp_file.readline())
    arr = list(map(int, inp_file.readline().split()))
    queries = int(inp_file.readline())
    for i in range(queries):
        k = int(inp_file.readline())
        kth_smallest_element = kth_smallest_value(arr, k)
        out_file.write(f"{kth_smallest_element}\n")
```
