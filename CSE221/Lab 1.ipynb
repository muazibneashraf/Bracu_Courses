```python
"""task1b.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/1Z6bOEe6c-12fwGa1GeY8xmlMeUZo4Tro
"""

input = open('/content/input1b.txt', 'r')
output= open('/content/output1b.txt', 'w')
temp = input.readline()
x = int(temp)
for i in range(x):
  data = input.readline()
  data_split = data.split()
  if data_split[2] == "+":
    output.write(f'The result of {data_split[1]} {data_split[2]} {data_split[3]} is {int(data_split[1]) + int(data_split[3])}\n')
  elif data_split[2] == "-":
    output.write(f'The result of {data_split[1]} {data_split[2]} {data_split[3]} is {int(data_split[1]) - int(data_split[3])}\n')
  elif data_split[2] == "*":
    output.write(f'The result of {data_split[1]} {data_split[2]} {data_split[3]} is {int(data_split[1]) * int(data_split[3])}\n')
  elif data_split[2] == "/":
    output.write(f'The result of {data_split[1]} {data_split[2]} {data_split[3]} is {int(data_split[1]) / int(data_split[3])}\n')
output.close()
```

```python
# -*- coding: utf-8 -*-
"""task2.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/1Z6bOEe6c-12fwGa1GeY8xmlMeUZo4Tro
"""

input = open('/content/input2.txt', 'r')
output= open('/content/output2.txt', 'w')
lenn = int(input.readline())
x = input.readline()
b = x.split()
arr = b
def bubbleSort(arr):
  for i in range(lenn - 1):
    #using flag for o(n)
    flag = False
    for j in range(lenn - i - 1):
      if arr[j] >= arr[j+1]:
        arr[j], arr[j+1] = arr[j+1], arr[j]
        flag = True
    if flag == False:
      break
  return arr
ans = bubbleSort(arr)
z = ""
for i in ans:
  z += i + " "
output.write(z)
output.close()
```

```python
# -*- coding: utf-8 -*-
"""task3.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/1Z6bOEe6c-12fwGa1GeY8xmlMeUZo4Tro
"""

input = open('/content/input3.txt', 'r')
output= open('/content/output3.txt', 'w')

arr_size=int(input.readline())
arr=[None]*(arr_size)
i=input.readline()
m=input.readline()

id=i.split()
mark=m.split()
for i in range(arr_size):
  arr[i]=((id[i],mark[i]))
for i in range(arr_size-1):
  min_ind=i
  for j in range(i+1,arr_size):
    if arr[j][1]>arr[min_ind][1] or (arr[j][1]==arr[min_ind][1] and arr[j][0]<arr[min_ind][0]):
      min_ind=j
  arr[i], arr[min_ind] = arr[min_ind],arr[i]

for i in arr:
  z='ID: '+i[0]+' Mark: '+i[1]+'\n'
  output.write(str(z))
output.close()
```

```python
# -*- coding: utf-8 -*-
"""task4.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/1Z6bOEe6c-12fwGa1GeY8xmlMeUZo4Tro
"""

input = open('/content/input4.txt', 'r')
output = open('/content/output4.txt', 'w')

data_num = int(input.readline())
data_list = input.readlines()
start = []
stop = []
time = []

for i in data_list:
  a = i.split()
  start.append(a[0])
  stop.append(a[4])
  time.append(a[-1])
for i in range(data_num):
  for j in range(0, data_num - i - 1):
    if start[j] > start[j + 1] or (start[j] == start[j + 1] and time[j] < time[j + 1]):
      start[j], start[j + 1] = start[j + 1], start[j]
      stop[j], stop[j + 1] = stop[j + 1], stop[j]
      time[j], time[j + 1] = time[j + 1], time[j]

for i in range(data_num):
  result = start[i] + ' will departure for ' + stop[i] + ' at ' + time[i] + '\n'
  output.write(str(result))
output.close()
```
