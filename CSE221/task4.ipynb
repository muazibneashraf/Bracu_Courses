```python
def min_coins(coins, target_amount):
    dp = [float('inf')] * (target_amount + 1)
    dp[0] = 0

    for coin in coins:
        for amount in range(coin, target_amount + 1):
            dp[amount] = min(dp[amount], dp[amount - coin] + 1)

    if dp[target_amount] != float('inf') :
      return dp[target_amount]
    else:
      return -1


with open("input4.txt", "r") as inp_file, open("output4.txt", "w") as out_file:
  n, target_amount = list(map(int, inp_file.readline().split()))
  coins = list(map(int, inp_file.readline().split()))
  out_file.write(f"{min_coins(coins, target_amount)}")
```
