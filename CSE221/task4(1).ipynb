```python
input=open("/content/input4.txt","r")
output=open("/content/output4.txt","w")


l=[int(i) for i in input.readline().strip().split(" ")]
num=l[0]
worker=l[1]
li=[]
for i in range(num):
    arr= [int(i) for i in input.readline().strip().split(" ")]
    li.append(arr)


def merge_sort(arr):
    if len(arr)<=1:
        return arr

    mid=len(arr)// 2
    left=arr[:mid]
    right=arr[mid:]

    left=merge_sort(left)
    right=merge_sort(right)
    return merge(left,right)

def merge(left,right):
    li_final=[]
    i=j=0
    while i<len(left) and j<len(right):
        if left[i][1]<right[j][1]:
            li_final.append(left[i])
            i+=1
        else:
            li_final.append(right[j])
            j+=1
    while i<len(left):
            li_final.append(left[i])
            i+=1
    while j<len(right):
            li_final.append(right[j])
            j+=1

    return li_final
li_ini=merge_sort(li)


li_ini=merge_sort(li)

count=0
for k in range(worker):
  prev_slot=li_ini[0]
  count+=1
  for i in range(len(li_ini)):
      if li_ini[i]!=0:
          if li_ini[i][0] >= prev_slot[1]:
              prev_slot=li_ini[i]
              count+=1
              li_ini[i]=0

output.write(str(count))

output.close()
```
