```python
def dfs(grid, row, col):
    stack = [(row, col)]
    diamonds_collected = 0
    while stack:
        r, c = stack.pop()
        if r < 0 or r >= len(grid) or c < 0 or c >= len(grid[0]) or grid[r][c] == '#' or grid[r][c] == 'X':
            continue
        if grid[r][c] == 'D':
            diamonds_collected += 1
            grid[r][c] = 'X'
        directions = [(0, 1), (0, -1), (1, 0), (-1, 0)]
        for dr, dc in directions:
            new_row, new_col = r + dr, c + dc
            stack.append((new_row, new_col))
            grid[r][c] = 'X'
    return diamonds_collected

with open("input6.txt", "r") as inp_file, open("output6.txt", "w") as out_file:
    R, H = map(int, inp_file.readline().split())
    grid = [list(inp_file.readline().strip()) for _ in range(R)]
    max_diamonds = 0
    for row in range(len(grid)):
        for col in range(len(grid[0])):
            if grid[row][col] == '.':
                max_diamonds = max(max_diamonds, dfs([row[:] for row in grid], row, col))
    out_file.write(f"{max_diamonds}")
```
