```python
# -*- coding: utf-8 -*-
"""task2b.ipynb

Automatically generated by Colaboratory.

Original file is located at
    https://colab.research.google.com/drive/160BfmDM-dz_56yxfishhYAzcFG9Tajio
"""

input=open("/content/input2b.txt","r")
output=open("/content/output2b.txt","w")


x=input.readline()
li_1=[int(i) for i in input.readline().strip().split(" ")]
y=input.readline()
li_2=[int(i) for i in input.readline().strip().split(" ")]

li_3=[]

def merge(left,right):
    li_final=[]
    i=0
    j=0
    while i<=len(left)-1 and j<=len(right)-1:
        if left[i]<right[j]:
            li_final.append(left[i])
            i+=1
        else:
            li_final.append(right[j])
            j+=1
    while i<=len(left)-1:
            li_final.append(left[i])
            i+=1
    while j<=len(right)-1:
            li_final.append(right[j])
            j+=1

    return li_final

li_3=merge(li_1,li_2)               # As the lists are already sorted only merge is used using a while loop , So the time complexity is O(n)

for i in range(len(li_3)):          # To trace the outputs 1 for loop is used, So the time complexity is O(n)
    output.write(str(li_3[i])+" ")
output.close()                      # The total time complexity is O(n +n) = O(n)
```
