```python
txt_input=open('/content/input1a.txt','r')
txt_output=open('/content/output1a.txt','w')
test=txt_input.readline().split()
r = int(test[0])
c = int(test[1])
a=[[0]*(r+1) for i in range(r+1)]
b=[]
for i in range(c):
  b.append([int(i) for i in txt_input.readline().split()])
#print(b)

for i in b:
  m, n, l = i
  a[m][n] = l
for i in a:
  #print(i)
  print(' '.join([str(j) for j in i]), file=txt_output)

txt_output.close()
```

```python
txt_input=open('/content/input1b.txt','r')
txt_output=open('/content/output1b.txt','w')
V,E= map(int, txt_input.readline().split())
List1 =[[]for i in range(0, V+1)]

for i in range(E):
  m, n, l = map(int, txt_input.readline().split())
  List1[m].append(tuple((n,l)))

for i in range(len(List1)):
  j = List1[i]
  txt_output.write(f"{i} :")
  for k,l in j:
    txt_output.write(f' {(k,l)}')
  txt_output.write("\n")

txt_output.close()
```
