```python
def alphanumeric(char):
  return 'a' <= char <= 'z' or '0' <= char <= '9'

def palindrome(s):
  s = s.lower()
  mu = [char for char in s if alphanumeric(char)]

  l, r = 0, len(mu) - 1
  while l < r:
    if mu[l] != mu[r]:
      return False

    l += 1
    r -= 1
  return True

example = "A man, a plan, a canal: Panama"
output = palindrome(example)
print(output)
```
