```python
from math import cos as c
from math import sin as s
import math
import numpy as np

np.set_printoptions(suppress=True)

def dh_matrix(a, d, alpha, theta):
    dh = [
        [c(theta), -c(alpha) * s(theta), s(alpha) * s(theta), a * c(theta)],
        [s(theta), c(alpha) * c(theta), -s(alpha) * c(theta), a * s(theta)],
        [0, s(alpha), c(alpha), d],
        [0, 0, 0, 1]]
    return np.array(dh)

T1 = dh_matrix(2, 8, math.radians(0), math.radians(30))
T2 = dh_matrix(7, 0, math.radians(0), math.radians(45))
T3 = dh_matrix(6, 0, math.radians(0), math.radians(20))
T4 = dh_matrix(0, 0, math.radians(90), math.radians(60))
T5 = dh_matrix(0, 3, math.radians(0), math.radians(30))

T = np.dot(np.dot(np.dot(np.dot(T1, T2), T3), T4), T5)
x = T[0, 3]
y = T[1, 3]
z = T[2, 3]

print("Forward solution:")
print(f"T1: \n{T1} \nT2: \n{T2} \nT3: \n{T3} \nT4: \n{T4} \nT5: \n{T5}")
print("Overall")
print(T)
print(f"Final position (x, y, z) of the end effector: ({x:.2f}, {y:.2f}, {z:.2f})")
```

```python
import math

def calculate_angles(x, y, z, l1, l2):
    theta1 = math.degrees(math.atan2(y, x))
    r = math.sqrt(x**2 + y**2)

    d2 = z
    d3 = math.sqrt(r**2 + d2**2)

    cos_theta2_a = (l1**2 + d3**2 - l2**2) / (2 * l1 * d3)
    theta2_a = math.degrees(math.acos(cos_theta2_a))

    tan_theta2_b = d2 / r
    theta2_b = math.degrees(math.atan(tan_theta2_b))

    theta2 = theta2_a + theta2_b

    cos_theta3 = (l1**2 + l2**2 - d3**2) / (2 * l1 * l2)
    theta3 = 180 - math.degrees(math.acos(cos_theta3))

    return theta1, theta2, theta3

x = 10.00
y = 28.00
z = 29.00
l1 = 48
l2 = 23

theta1, theta2, theta3 = calculate_angles(x, y, z, l1, l2)
print(f"theta 1: {theta1:.2f}")
print(f"theta 2: {theta2:.3f}")
print(f"theta 3: {theta3:.2f}")
```
